## 簡介

**作者註**：這一段是從 g-h 濾波器那一章剪過來的，它本來就不該放在那裡。這一章還不是給人讀的！我還沒寫。

在那一章的開頭附近，我用 `numpy.polyfit()` 對體重量測值擬合了一條直線。它使用「least squared fit（最小平方擬合）」對資料擬合一個 n 次多項式。這和 g-h filter（g-h 濾波器）有什麼不同？

嗯，要看情況。我們最終會學到，在某些條件下 Kalman filter（卡爾曼濾波器）從最小平方擬合的觀點來看是最佳的。然而，`polyfit()` 是對資料擬合一個多項式，而不是任意曲線，方法是把下面這個式子的值最小化：

$$E = \sum_{j=0}^k |p(x_j) - y_j|^2$$

我假設我的體重以每天 1 lb 的固定速率增加，所以當我嘗試擬合 $n=1$ 的多項式（也就是一條直線）時，結果與實際的體重增加非常吻合。但是，當然，沒有人會持續只增重或只減重。我們會上下波動。對更長的一串資料使用 'polyfit()' 會得到很差的結果。相較之下，g-h filter 會對速率的變化做出反應——$h$ 這一項控制濾波器對這些變化反應得多快。如果我們增重、維持一陣子、然後減重，濾波器會自動追蹤那個變化。'polyfit()' 做不到這一點，除非增重與減重可以用一個多項式很好地表示。

這種形式的濾波器還有另一個優點，即使資料真的符合某個 *n* 次多項式也一樣：它是 *recursive*（遞迴）的。也就是說，我們只需要知道上一個時間週期的估計值與速率，就能計算出這個時間週期的估計值，其他什麼都不需要。相較之下，如果你深入 `polyfit()` 的實作，你會看到它需要拿到全部的資料之後才能給出答案。因此，像 `polyfit()` 這類演算法並不適合即時的資料濾波。在 1960 年代 Kalman filter 被發展出來的時候，電腦非常慢，記憶體也極為有限。它們完全無法儲存例如飛機 inertial navigation system（慣性導航系統）的數千筆讀數，也無法在提供準確且即時的導航資訊所需的短暫時間內處理完那麼多資料。 


直到 20 世紀中葉，這類濾波都是使用各種形式的 Least Squares Estimation（最小平方估計）。舉例來說，NASA 的 Apollo 計畫有一個地面網路用來追蹤 Command and Service Module（CSM，指揮服務艙）與 Lunar Module（LM，登月艙）。他們花很多分鐘取得量測值，把資料批次整理在一起，再慢慢算出答案。1960 年，NASA Ames 的 Stanley Schmidt 看出了 Rudolf Kalman 那篇開創性論文的實用價值，便邀請他到 Ames。Schmidt 把 Kalman 的成果應用到 CSM 與 LM 的機載導航系統上，並稱之為「Kalman filter」。[1] 不久之後，全世界都轉向了這種更快速的遞迴濾波器。

Kalman filter 只需要儲存上一次的估計值和幾個相關參數，而且只需要相對少量的計算就能產生下一個估計值。今天我們有這麼多記憶體與運算能力，這項優勢的重要性已經降低了一些，但在當時 Kalman filter 是一項重大突破，不只是因為它的數學性質，也因為它（勉強）能在當年的硬體上執行。
