# 安裝 Installation

本書是用 Jupyter Notebook 寫成的，這是一個以瀏覽器為基礎的互動式 Python 環境，可以把 Python、文字與數學混在一起。我選擇它是因為它的互動特性——在我開始在互動式環境裡工作之前，我發現 Kalman filtering（卡爾曼濾波）幾乎不可能學得會。在你能夠親手改變參數並立刻看到輸出之前，很難對許多參數建立直覺。互動式環境也讓你可以玩「如果……會怎樣」的情境。「如果我把 $\mathbf{Q}$ 設成零會怎樣？」用 Jupyter Notebook 要找出答案輕而易舉。

## SymPy

SymPy 是一個用來進行符號數學運算的 Python 套件。它的完整能力超出本書的範圍，但它可以做代數運算、對方程式做積分與微分、求 differential equation（微分方程）的解，還有更多功能。舉例來說，我們用它來計算矩陣的 Jacobian（雅可比矩陣）以及 expected value（期望值）的積分計算。

首先來個簡單的例子。我們會 import SymPy，初始化它的 pretty print 功能（它會用 LaTeX 印出方程式）。接著我們宣告一個符號給 SymPy 使用。

```python
import sympy
sympy.init_printing(use_latex='mathjax')

phi, x = sympy.symbols('\phi, x')
phi
```

注意它是如何用 LaTeX 印出符號 `phi` 的。現在來做點數學。$\sqrt{\phi}$ 的導數是什麼？

```python
sympy.diff('sqrt(phi)')
```

我們可以對方程式做因式分解

```python
sympy.factor(phi**3 -phi**2 + phi - 1)
```

也可以把它們展開。

```python
((phi+1)*(phi-4)).expand()
```

你可以用變數的特定數值來求方程式的值：

```python
w =x**2 -3*x +4
print(w.subs(x, 4))
print(w.subs(x, 12))
```

對於使用了你尚未定義之符號的方程式，你也可以用字串來表示：

```python
x = sympy.expand('(t+1)*2')
x
```

現在讓我們用 SymPy 來計算一個矩陣的 Jacobian。給定函數

$$h=\sqrt{(x^2 + z^2)}$$

求它對 x、y、z 的 Jacobian。

```python
x, y, z = sympy.symbols('x y z')

H = sympy.Matrix([sympy.sqrt(x**2 + z**2)])

state = sympy.Matrix([x, y, z])
H.jacobian(state)
```

現在讓我們根據連續的 process noise（過程雜訊）矩陣 
$$\mathbf Q = \Phi_s \begin{bmatrix}0&0&0\\0&0&0\\0&0&1\end{bmatrix}$$

來計算離散的 process noise 矩陣 $\mathbf Q$。積分為 

$$\mathbf Q = \int_0^{\Delta t} \mathbf F(t)\mathbf Q\mathbf F^T(t)\, dt$$

其中 
$$\mathbf F(\Delta t) = \begin{bmatrix}1 & \Delta t & {\Delta t}^2/2 \\ 0 & 1 & \Delta t\\ 0& 0& 1\end{bmatrix}$$

```python
dt = sympy.symbols('\Delta{t}')
F_k = sympy.Matrix([[1, dt, dt**2/2],
                    [0,  1,      dt],
                    [0,  0,       1]])
Q = sympy.Matrix([[0,0,0],
                  [0,0,0],
                  [0,0,1]])

sympy.integrate(F_k*Q*F_k.T,(dt, 0, dt))
```
