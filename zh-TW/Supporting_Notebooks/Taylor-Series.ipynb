# 用 Taylor Series（泰勒級數）做線性化 Linearizing with Taylor Series

Taylor series 把一個函數表示成無窮多項的總和。即使對 nonlinear（非線性）函數而言，這些項也是 linear（線性）的，所以我們可以用線性代數來表示任意的非線性函數。這個選擇的代價是：除非使用無窮多項，否則我們算出的值只會是近似值，而非精確值。

實函數或複函數 f(x) 在 x=a 處的 Taylor series 是以下的無窮級數

$$f(x) = f(a) + f'(a)(x-a) + \frac{f''(a)}{2!}(x-a)^2 + \, ...\,  + \frac{f^{(n)}(a)}{n!}(x-a)^n + \, ...$$

其中 $f^{n}$ 是 f 的 n 階 derivative（導數）。為了計算 $f(x)=sin(x)$ 在 $x=0$ 處的 Taylor series，我們先算出 f 的各項。

$$\begin{aligned}
f^{0}(x) &= sin(x) ,\ \  &f^{0}(0) &= 0 \\
f^{1}(x) &= cos(x),\ \  &f^{1}(0) &= 1 \\
f^{2}(x) &= -sin(x),\ \  &f^{2}(0) &= 0 \\
f^{3}(x) &= -cos(x),\ \  &f^{3}(0) &= -1 \\
f^{4}(x) &= sin(x),\ \  &f^{4}(0) &= 0 \\
f^{5}(x) &= cos(x),\ \  &f^{5}(0) &= 1
\end{aligned}
$$

現在我們可以把這些值代入方程式。

$$\sin(x) = \frac{0}{0!}(x)^0 + \frac{1}{1!}(x)^1 + \frac{0}{2!}(x)^2 + \frac{-1}{3!}(x)^3 + \frac{0}{4!}(x)^4 + \frac{1}{5!}(x)^5 + ... $$

接著用一些程式碼來測試：

```python
import numpy as np

x = .3
estimate = x - x**3/6 + x**5/120
exact = np.sin(.3)

print('estimate of sin(.3) is', estimate)
print('exact value of sin(.3) is', exact)
```

只用三項就有這樣的結果，還不錯。如果你感到好奇，不妨把它實作成一個 Python 函式，計算任意項數的級數。
