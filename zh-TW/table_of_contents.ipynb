我們把在 univariate（單變量）那一章發展出的 Kalman filter，擴展成適用於 linear（線性）問題的完整、一般化濾波器。讀完之後，你會了解 Kalman filter 如何運作，以及如何針對你所選的（線性）問題設計並實作一個 Kalman filter。

[**Chapter 9: Nonlinear Filtering（第 9 章：非線性濾波）**](./09-Nonlinear-Filtering.ipynb)

目前介紹的 Kalman filter 只適用於線性問題。然而這個世界是非線性的。這裡我會介紹非線性系統對濾波器帶來的問題，並簡要討論我們在後續章節將學到的各種演算法。


[**Chapter 10: Unscented Kalman Filters（第 10 章：無損卡爾曼濾波器）**](./10-Unscented-Kalman-Filter.ipynb)

Unscented Kalman filter（UKF，無損卡爾曼濾波器）是 Kalman filter 理論中較新的發展。它讓你能對非線性問題進行濾波，而不需要像 Extended Kalman filter（擴展卡爾曼濾波器）那樣要求 closed form（閉合形式）的解。

這個主題在現有教科書中通常不是完全沒提，就是一筆帶過，大部分篇幅都給了 Extended Kalman filter。我把它放在前面，是因為 UKF 理解和實作起來都簡單得多，而且濾波效能通常和 Extended Kalman filter 一樣好，甚至更好。面對真實世界的問題時，我總是先嘗試實作 UKF，你也應該這麼做。


[**Chapter 11: Extended Kalman Filters（第 11 章：擴展卡爾曼濾波器）**](./11-Extended-Kalman-Filters.ipynb)

Extended Kalman filter（EKF）是把非線性問題 linearization（線性化）最常見的做法。真實世界中大多數的 Kalman filter 都是 EKF，所以你需要了解這些內容，才能看懂既有的程式碼、論文、演講等等。 


[**Chapter 12: Particle Filters（第 12 章：粒子濾波器）**](./12-Particle-Filters.ipynb)

Particle filter（粒子濾波器）使用 Monte Carlo（蒙地卡羅）技術來對資料進行濾波。它能輕鬆處理高度非線性與非高斯的系統，以及多峰分布（同時追蹤多個物體），代價是很高的計算需求。


[**Chapter 13: Smoothing（第 13 章：平滑）**](./13-Smoothing.ipynb)

Kalman filter 是 recursive（遞迴）的，因此非常適合即時濾波。不過，它們在資料的後處理上也表現得極為出色。畢竟 Kalman filter 是預測－校正器（predictor-corrector），而預測過去總比預測未來容易！我們會討論一些常見的做法。


[**Chapter 14: Adaptive Filtering（第 14 章：適應性濾波）**](./14-Adaptive-Filtering.ipynb)
    
Kalman filter 假設只有單一個 process model（過程模型），但會機動轉向的目標通常需要用數個不同的過程模型來描述。Adaptive filtering（適應性濾波）運用多種技術，讓 Kalman filter 能適應目標不斷變化的行為。
